```python
import pandas as pd
df = pd.read_csv('data.csv')
df['city_dwarka'] =1
df['num_lanes'] = 2
```

```python
all_sumo_road_types = ['highway.residential','highway.tertiary','highway.secondary','highway.primary']
```

```python
# performing one hot encoding on the road_types
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
ohe = OneHotEncoder(categories=[all_sumo_road_types], sparse_output=False, handle_unknown='ignore', dtype=int)
road_encoded = ohe.fit_transform(df[['road_type']])
encoded_df = pd.DataFrame(road_encoded, columns=ohe.get_feature_names_out(['road_type']))
df = pd.concat([df.drop('road_type', axis=1), encoded_df], axis=1)
```

```python
df.sample(3)
```

```python
df.info()
```

```python
numerical_cols = [
    'length_m', 'speed_limit_mps', 'num_lanes', 'priority',
    'free_flow_travel_time_s', 'from_junction_degree', 'to_junction_degree',
    'baseline_sampled_seconds', 'baseline_traveltime_s', 'baseline_overlap_traveltime_s',
    'baseline_density_vpm', 'baseline_lane_density_vpmpl', 'baseline_occupancy_perc',
    'baseline_waiting_time_s', 'baseline_time_loss_s', 'baseline_speed_mps',
    'baseline_speed_relative', 'baseline_departed_count', 'baseline_arrived_count',
    'baseline_entered_count', 'baseline_left_count'
]
```

```python
#performing the feature scaling of all numerical columns using StandardScaler
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
df.loc[:,numerical_cols] = scaler.fit_transform(df[numerical_cols])
```

```python
df
```

```python
from sklearn.model_selection import train_test_split
x = df.drop(['edge_id', 'from_node_id', 'to_node_id', 'is_braess_road'], axis=1)
y = df['is_braess_road']
x_train , x_test  ,y_train , y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train
```

```python
# Training the Model
from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier(n_estimators = 100,random_state = 42)
rf.fit(x_train,y_train)
y_pred = rf.predict(x_test)
y_pred_proba = rf.predict_proba(x_test)[:, 1] # Probabilities for the positive class
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, confusion_matrix
print(f"Accuracy: {accuracy_score(y_test, y_pred):.4f}")
print(f"Precision: {precision_score(y_test, y_pred):.4f}")
print(f"Recall: {recall_score(y_test, y_pred):.4f}")
print(f"F1-Score: {f1_score(y_test, y_pred):.4f}")
print(f"ROC AUC Score: {roc_auc_score(y_test, y_pred_proba):.4f}")
```
